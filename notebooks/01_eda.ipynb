```python
gt["has_match"] = gt["matched_entity_ids"].fillna("").str.strip().ne("")

print(gt["has_match"].value_counts())
```

```python
gt["match_count"] = gt["matched_entity_ids"].fillna("").apply(
    lambda x: 0 if not x else len(x.split(","))
)

print(gt["match_count"].describe())
```

```python
print(
    gt["match_count"]
    .value_counts()
    .sort_index()
    .head(20)
)
```

```python
def count_sources(x):
    if not x:
        return 0, 0

    ids = x.split(",")

    s2_count = sum(i.startswith("S2-") for i in ids)
    s3_count = sum(i.startswith("S3-") for i in ids)

    return s2_count, s3_count


gt[["s2_count", "s3_count"]] = gt["matched_entity_ids"].fillna("").apply(
    lambda x: pd.Series(count_sources(x))
)

print(gt[["s2_count", "s3_count"]].describe())
```

```python
print("S2 matches:", (gt["s2_count"] > 0).sum())
print("S3 matches:", (gt["s3_count"] > 0).sum())
print("Both S2 & S3:", ((gt["s2_count"] > 0) & (gt["s3_count"] > 0)).sum())
```
